```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data=pd.read_csv("bank_note_data.csv")
```

```python
data.head()
```

```python
feature=list(data.columns)[:-1]
```

```python
feature
```

```python
target=list(data.columns)[-1]
```

```python
target
```

## Scaling Data

```python
from sklearn.preprocessing  import StandardScaler
data_norm=data.copy()
sc=StandardScaler()
data_norm[feature]=sc.fit_transform(data[feature])
```

```python
data_norm
```

# TSNE

```python
from sklearn.manifold import TSNE

tsne=TSNE(learning_rate=500,n_components=2)

x_tsne=tsne.fit_transform(data_norm[feature])
```

```python
x_tsne
```

```python
y=data[target]
```

```python
y
```

## Visulization

```python
plt.figure(figsize=(12,10))

plt.scatter(x_tsne[np.where(y==0),0],x_tsne[np.where(y==0),1],color='g',marker='*')
plt.scatter(x_tsne[np.where(y==1),0],x_tsne[np.where(y==1),1],color='r',marker='+')

plt.xlabel("Dimension_1")
plt.ylabel("Dimension_2")
plt.title("TSNE PLOT")
```
