```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.axes as ax

from sklearn import datasets
```

```python
data=pd.read_csv("boston_house_prices.csv")
```

```python
data.head()
```

```python
data=data.rename(columns={'506':"CRIM","13":"ZN","Unnamed: 2":"INDUS","Unnamed: 3":"CHAS","Unnamed: 4":"NOX","Unnamed: 5":"RM","Unnamed: 6":"AGE","Unnamed: 7":"DIS","Unnamed: 8":"RAD",
                          "Unnamed: 9":"TAX","Unnamed: 10":"PTRATIO","Unnamed: 11":"B","Unnamed: 12":"LSTAT","Unnamed: 13":"MEDV"})
```

```python
data=data.drop(index=0)
```

```python
data=data.reset_index(drop=True)
data.head()
```

```python
x=data.drop(["MEDV"],axis=1)
y=data["MEDV"]
```

```python
x
```

```python
y
```

```python
from sklearn.preprocessing import StandardScaler

sc=StandardScaler()
x=sc.fit_transform(x)
```

```python
x
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
```

```python
x_train.shape
```

```python
y_test.shape
```

## Random Forest

```python
from sklearn.ensemble import RandomForestRegressor
rf=RandomForestRegressor(100)
rf=rf.fit(x_train,y_train)

pred=rf.predict(x_test)
```

```python
pred.shape
```

```python
from sklearn.metrics import mean_squared_error
mean_squared_error(y_test,pred)
```

## K_Fold Cross validation

```python
from sklearn.model_selection import KFold
kfold=KFold(n_splits=5,shuffle=False)

for train_index, test_index in kfold.split(x):
    print(train_index,"\n\n",test_index)
```

```python
rf1=RandomForestRegressor()
error=[]

for train_index, test_index in kfold.split(x):
    x_train,x_test=x[train_index],x[test_index]
    y_train,y_test=y[train_index],y[test_index]

    rf1=RandomForestRegressor(100)

    rf1=rf1.fit(x_train,y_train)

    pred=rf1.predict(x_test)
    error.append(mean_squared_error(y_test,pred))
```
